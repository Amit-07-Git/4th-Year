```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0,10,100)
y = x**2
fig,ax = plt.subplots()
fig.set_facecolor('blue')
ax.set_facecolor('yellow')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('y=x2')
ax.grid(color='red', linestyle='dotted', linewidth=0.3)
ax.plot(x,y,'green',lw=3)
plt.show()

print(fig.get_figwidth())
print(fig.get_figheight())
# Default resolution of figure size is 640 x 480
```

```python
y1 = x**2
y2 = 2*x
fig = plt.figure()
ax1 = fig.add_axes([0.2,0.2,0.3,0.3])
ax1.plot(x,y1,label='y1=x^2')
ax1.plot(x,y2,label='y2=2x')
ax1.legend()
plt.show()
```

```python
#WAPP to draw a func u = 325*sin(2*pie*f*t) where f = 50hz t = 20ms identify the peak of the function and the function line using legend
f = 50
t = np.linspace(0,0.02,100)
u = 325 * np.sin(2 * np.pi * f * t)
fig, ax = plt.subplots()
ax.plot(t * 1000, u, 'blue', lw=3, label='u = 325 * sin(2 * np.pi * f * t)')
ax.plot(5, 325, 'ro', label = 'Peak = 325 V')
plt.plot(t * 1000, u, color='blue')
plt.title('Voltage u(t)')
plt.xlabel('Time (ms)')
plt.ylabel('Voltage (V)')
ax.legend()
plt.show()
```

```python
t = np.linspace(0,20,1000)
u = 325 * np.sin(2 * np.pi * t / 20)

fig, ax = plt.subplots()
ax.plot(t , u)
ax.annotate('Peak value = 325 V',
            xy = (5, 325),
            xytext = (8, 280),
            arrowprops = dict(arrowstyle = '->',))
ax.set_xlabel('Time (ms)')
ax.set_ylabel('Voltage (V)')
plt.show()
```

```python
#WAPP to annotate 3 func lines as R1, R2, R3 where R1,R2,R3 are resistors 2, 4, 8 respectively and you have to calculate u1, u2, u3 and plot the current i vs u1, u2, u3 i will be in the range to 0.1 to 1 create 100 values within this range. (U = R X i)
R1, R2, R3 = 2, 4, 8
I = np.linspace(0.1,1,100)
u1 = R1 * I
u2 = R2 * I
u3 = R3 * I
fig, ax = plt.subplots()
ax.plot(I, u1, 'blue', label='u1 = I')
ax.plot(I, u2, 'green', label='u2 = I')
ax.plot(I, u3, 'red', label='u3 = I')
ax.annotate('R1', (I[70], u1[70]))
ax.annotate('R2', (I[70], u2[70]))
ax.annotate('R3', (I[70], u3[70]))
ax.legend()
plt.show()
```
